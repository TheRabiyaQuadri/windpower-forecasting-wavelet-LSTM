```python
import pandas as pd
import numpy as np
```

```python
import csv

# Define input and output file paths
input_file = "Data B_Li.txt"  # Replace with your text file path
output_file = "output.csv"

# Open the input file and read lines
with open(input_file, "r") as file:
    lines = file.readlines()

# Extract the headers and data rows
headers = lines[0].strip().split('\t')
data_rows = [line.strip().split('\t') for line in lines[1:]]

# Write the data to a CSV file
with open(output_file, "w", newline='') as csv_file:
    writer = csv.writer(csv_file)
    writer.writerow(headers)  # Write header row
    writer.writerows(data_rows)  # Write data rows

print(f"Data has been successfully converted to {output_file}")
```
